Iniciar uma sessão local e importar dados:

```python
# Iniciar uma sessão local e importar dados
from pyspark.sql import SparkSession
sc = SparkSession.builder.master('local[*]').getOrCreate()
```

Base de dados usada: https://www.kaggle.com/datasets/mahmoudelhemaly/students-grading-dataset?resource=download

```python

# carregar dados do meu computador
dados_spark = sc.read.csv("/content/Students_Grading_Dataset.csv", header=True, inferSchema=True)

# Verificando o tipo de objeto criado
type(dados_spark)
```

```python
dados_spark = dados_spark.withColumnRenamed("Student_ID", "ID_do_Aluno") \
    .withColumnRenamed("First_Name", "Primeiro_Nome") \
    .withColumnRenamed("Last_Name", "Sobrenome") \
    .withColumnRenamed("Email", "Email") \
    .withColumnRenamed("Gender", "Genero") \
    .withColumnRenamed("Age", "Idade") \
    .withColumnRenamed("Department", "Departamento") \
    .withColumnRenamed("Attendance (%)", "Presenca") \
    .withColumnRenamed("Midterm_Score", "Nota_Prova_Parcial") \
    .withColumnRenamed("Final_Score", "Nota_Prova_Final") \
    .withColumnRenamed("Assignments_Avg", "Media_Tarefas") \
    .withColumnRenamed("Quizzes_Avg", "Media_Quizzes") \
    .withColumnRenamed("Participation_Score", "Nota_Participacao") \
    .withColumnRenamed("Projects_Score", "Nota_Projetos") \
    .withColumnRenamed("Total_Score", "Nota_Total") \
    .withColumnRenamed("Grade", "Conceito") \
    .withColumnRenamed("Study_Hours_per_Week", "Horas_Estudo_Semana") \
    .withColumnRenamed("Extracurricular_Activities", "Atividades_Extracurriculares") \
    .withColumnRenamed("Internet_Access_at_Home", "Acesso_Internet_Casa") \
    .withColumnRenamed("Parent_Education_Level", "Nivel_Educacao_Pais") \
    .withColumnRenamed("Family_Income_Level", "Nivel_Renda_Familiar") \
    .withColumnRenamed("Stress_Level (1-10)", "Nivel_Estresse") \
    .withColumnRenamed("Sleep_Hours_per_Night", "Horas_Sono_Noite")
```

```python
# Verificando o dataset
dados_spark.show(5)
```

```python
# Verificando o schema() deste spark dataframe
dados_spark.printSchema()
```

```python
#Retornar o número de linhas
dados_spark.count()
```

```python
#Selecionar colunas
dados_selecionados = dados_spark.select("ID_do_Aluno","Genero","Idade","Departamento","Nota_Prova_Final").show(5)
```

```python
#Principais estatísticas
dados_spark.describe('Nota_Prova_Final').show(5)
```

```python
#Ocorrências em uma variável
dados_spark.select("Departamento","Genero").distinct().show()
```

```python
#importar sql: groupBy
from pyspark.sql import functions as F

dados_spark.groupBy("Departamento").agg(F.sum("Horas_Estudo_Semana")).show()
```

# **Quais foram as maiores médias por tarefas?**

```python
item_popular = dados_spark.groupBy('Media_Tarefas').agg(F.count('Media_Tarefas')).orderBy(F.count('Media_Tarefas').desc()).show(10)
```

#**Qual é a faixa etaria por depatamento?**

```python
# prompt: crie uma analise de faixa etária por departamento atraves do codigo a seguir

# Agrupar por departamento e calcular a faixa etária
faixa_etaria_por_departamento = dados_spark.groupBy("Departamento").agg(
    F.min("Idade").alias("Idade_Minima"),
    F.max("Idade").alias("Idade_Maxima"),
)

# Exibir os resultados
faixa_etaria_por_departamento.show()
```

# **Qual é o nível de frequencia por alunos?**

```python
gasto_cliente = dados_spark.groupBy('ID_do_Aluno').agg(F.count('ID_do_Aluno'),F.sum('Presenca')).orderBy(F.sum('Presenca').desc()).show()
```

```python
#Dados faltantes
dados_spark.select([F.count(F.when(F.isnull(c), c)).alias(c) for c in dados_spark.columns]).show()
```

```python
dados_spark.select('Nota_Participacao').distinct().show()
```

```python
#substituir valores ausentes
dados_spark_sem_na = dados_spark.fillna({'Nota_Participacao':0, 'Nota_Projetos':0})
```

```python
#Excluir colunas da base de dados
dados_spark = dados_spark.drop('Nota_Participacao','Nota_Projetos')
dados_spark.show(5)
```
